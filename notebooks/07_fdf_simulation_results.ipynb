```python
dfs = []
for x in os.listdir('../data/interim/simulation/20260416/cv/'):
    dfs.append(joblib.load(''.join(['../data/interim/simulation/20260416/cv/', x])))
df_cv = pd.concat(dfs)
```

```python
df_cv.sim_n_rep.unique()
```

```python
df_cv.groupby(["sim_base_pdf_name", "kde_model_name"])["KLD"].mean().sort_values()
```

```python
for scenario in df_cv.sim_scenario.unique():
    df_cv_2 = df_cv[df_cv["sim_scenario"]==scenario]
    sim_base_pdf = df_cv_2["sim_base_pdf_name"].unique()[0]
    for kde_model_name in df_cv_2.kde_model_name.unique():
        df_cv_3 = df_cv_2[df_cv_2["kde_model_name"]==kde_model_name]

        # plt.figure(figsize=(15,5))
        # for fc_date in df_cv_3.fold_fc_date.unique():
        #     df_cv_4 = df_cv_3[df_cv_3["fold_fc_date"] == fc_date].iloc[0,:]
        #     curves = df_cv_4["curves"].reset_index()
        #     curves = curves[["support", "actual", "forecast"]].set_index("support")
            
        #     plt.title(f"base pdf: {sim_base_pdf} | {kde_model_name}")
        #     plt.plot(curves["actual"], color="gray", alpha=0.5)
        #     plt.plot(curves["forecast"], color="blue", alpha=0.5)
        # plt.show()

        plt.figure(figsize=(15,5))
        for fc_date in df_cv_3.fold_fc_date.unique():
            df_cv_4 = df_cv_3[df_cv_3["fold_fc_date"] == fc_date].iloc[0,:]
            curves = df_cv_4["curves"].reset_index()
            curves = curves[["support", "actual", "forecast"]].set_index("support")
            curves["residuals"] = curves.actual - curves.forecast
            curves["rel_resid"] = np.abs(curves["residuals"])/curves.actual
            
            plt.title(f"base pdf: {sim_base_pdf} | {kde_model_name}")
            plt.plot(curves["rel_resid"], color="gray", alpha=0.5)
        plt.show()
```
