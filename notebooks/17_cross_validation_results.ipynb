## Cross-validation procedure

Let $\mathbb{K}$ denote the set of kernel functions and $\mathcal{B}(K)$ the set of bandwidth selection 
procedures available for kernel $K$. In addition, let $\mathbb{P}$ denote the set of post-processing configurations. 
Each $P \in \mathbb{P}$ specifies a rule for computing a local mean of the density curves and is defined by a 
mean-computation mode and an associated window length $w$. In the empirical application, the mean-computation mode 
may correspond to an expanding or rolling scheme, while the window length determines the number of past curves 
used in the computation. The set $\mathbb{P}$ also includes the configuration corresponding to no
post-processing, for which the transformation operator reduces to the
identity mapping $\mathcal{T}_{P}(f_t)=f_t$.


The complete parameter space is defined as
\begin{equation}
\Theta = \{(K, B, \eta, P) : K \in \mathbb{K},\; B \in \mathcal{B}(K),\; P \in \mathbb{P}\},
\end{equation}
where $\eta$ captures auxiliary parameters, such as the degrees of freedom for the Student-t kernel or the robust scale estimator 
for Silverman’s rule. In the present setting, these correspond to the degrees of freedom parameter for the 
Student-$t$ kernel or the use of a robust standard deviation estimator in 
the Silverman bandwidth rule.
Each configuration $\theta_j \in \Theta$ is therefore defined as
\begin{equation}
\theta_j = (K_j, B_j, \eta_j, P_j), \quad j = 1,\dots,M.
\end{equation}

Let $r_{t1},\dots,r_{tn_t}$ denote the intraday log-returns observed on day $t$, 
for $t=1,\dots,T$. For a given configuration $(K_j,B_j,\eta_j)$, the raw daily density estimates are first obtained using the kernel density estimator


\begin{equation}
f_t^{(K_j,B_j,\eta_j)}(x)
=
\frac{1}{n_t h_t^{(j)}}
\sum_{i=1}^{n_t}
K_j\!\left(\frac{r_{ti}-x}{h_t^{(j)}};\eta_j\right),
\end{equation}

where $h_t^{(j)}$ is the bandwidth determined by the rule $B_j$.
This produces the functional time series of densities
\begin{equation}
\mathcal{X}^{(K_j,B_j,\eta_j)}=\{f_1^{(K_j,B_j,\eta_j)},\dots,f_T^{(K_j,B_j,\eta_j)}\}.
\end{equation}

The post-processing configuration $P_j$ is then applied through a transformation operator $\mathcal{T}_{P_j}$ acting on the sequence of density curves. This yields the transformed functional time series
\begin{align*}
\mathcal{Z}^{(\theta_j)} &= \{Z_1^{(\theta_j)},\dots,Z_T^{(\theta_j)}\}, \\
Z_t^{(\theta_j)}         &= \big(\mathcal{T}_{P_j}(\mathcal{X}^{(K_j,B_j,\eta_j)})\big)_t.
\end{align*}

The sample is partitioned into three segments: an initial estimation period,
a validation period, and a test period. Let $t_0$ denote the size of the
initial estimation window and let $t_v$ denote the final index of the
validation period, where $t_0 < t_v < T$. The observations
$t=1,\dots,t_0$ form the initial estimation sample. Forecasts corresponding
to $t=t_0+1,\dots,t_v$ constitute the validation period used for model
selection, while forecasts for $t=t_v+1,\dots,T$ form the test period used
for final forecast evaluation.
For each configuration $\theta_i$, forecasts are generated using an
expanding window scheme. At each step $t=t_0,\dots,T-1$, the forecasting
model is estimated using the sample
${Z_1^{(\theta_i)},\dots,Z_t^{(\theta_i)}}$ and produces a one-step-ahead
forecast $\widehat{Z}_{t+1}^{(\theta_i)}$. This procedure yields the sequence
$
\widehat{Z}_{t_0+1}^{(\theta_i)},\widehat{Z}_{t_0+2}^{(\theta_i)},\dots,
\widehat{Z}_{T}^{(\theta_i)}.
$
In the empirical application considered in this dissertation, the initial
estimation window is set to $t_0=100$, and the validation period ends at
$t_v=180$. Consequently, the validation and test partitions contain
80 and 67 evaluation curves, respectively.

\begin{equation}
L_{ij}
=
\frac{1}{t_v - t_0}
\sum_{t=t_0}^{t_v-1}
D\!\left(
Z_{t+1}^{(\theta_j)},
\widehat{Z}_{t+1}^{(\theta_i)}
\right).
\end{equation}

which yields an $M \times M$ loss matrix. Finally, the selected configuration is


\begin{equation*}
\theta^*
=
\arg\min_{\theta_i \in \Theta}
\frac{1}{M}
\sum_{j=1}^{M}
L_{ij}.
\end{equation*}

* Kernel options:
    - gaussian
    - epanechnikov
    - student-t
* Bandwidth selection method:
    - ROT
    - CV
    - Adaptive
* Postprocessing configurations:
    - No postprocessing (False)
    - 5 days (w5)
    - 21 days (w21)
    - 63 days (w63)

In out configuration, M = 53.

```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

import joblib
```

```python
df_cv = joblib.load('../data/processed/cv_divergences_long_20260316.jbl')
df_cv["forecast_kde_pp_params"] = df_cv["forecast_kde_params"] + "___" + df_cv["forecast_pp_params"]
display(df_cv.head())
df_cv.shape
```

* kde_model: the database $\theta_j$
* forecast_kde_pp_params: kernel, bandwidth and postprocessing params for forecasting with $\theta_i$
* KLD, JSD, L-norms = $D(\theta_{tj}, \theta_{ti})$

```python
# Top models by kernel + bandwidth selection method
df_cv.groupby("forecast_kde_params")["KLD"].mean().sort_values()
```

```python
# Top models by postprocessing method
df_cv.groupby("forecast_pp_params")["KLD"].mean().sort_values()
```

```python
criterion = "forecast_pp_params"

results_by_criterion = df_cv.loc[:, [criterion, 'KLD', 'JSD', 'L1_norm', 'L2_norm', 'LINF_norm']]
rbc_long = results_by_criterion.melt(id_vars=[criterion], var_name="metric")
rbc_long[criterion] = rbc_long.loc[:,criterion].apply(str)

if rbc_long[criterion].nunique() >= len(['KLD', 'JSD', 'L1_norm', 'L2_norm', 'LINF_norm']):
    columns_by = None
    rows_by   = "metric"
    aspect=3
else:
    columns_by = "metric"
    rows_by   =  None
    aspect=1

plt.figure(figsize=(10, 5))

print(criterion)
g = sns.catplot(
    data=rbc_long,
    x=criterion,
    y="value",
    row=rows_by,
    col=columns_by,
    kind="box",
    hue=criterion,
    sharey=False,
    sharex=False,
    height=3,
    aspect=aspect,
    saturation=.5,
    showfliers=False,
)

plt.suptitle(f"Results by {criterion}",y=1.03)

plt.show()
```

```python
# Top models by kernel + bandwidth selection + postprocessing method
df_cv.groupby("forecast_kde_pp_params")["KLD"].mean().sort_values()
```
