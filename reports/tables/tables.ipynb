```python
import numpy as np
import pandas as pd

tables_path = "../../data/processed/"
out_path    = "../../../densities4risk_doc/Tables/"
```

# Cross validation

```python
cv_results = pd.read_excel('/'.join([tables_path, 'cv/20260225/','cv_density_forecasting_20260224.xlsx']))
cv_results = cv_results[['normalized_density', 'bandwidth', 'kernel', 'robust_std', #'dFPC_dimensions',
                          'KLD', 'JSD', 'L_1', 'L_2', 'L_INFTY']]
cv_results_summary = cv_results.groupby(['normalized_density', 'bandwidth', 'kernel', 'robust_std'], as_index=False).mean()
cv_results.sort_values(by="KLD")
```

```python
# cv_results = pd.read_excel('/'.join([tables_path, 'cv_density_estimation_v3.xlsx']))#.drop(columns=["Unnamed: 0"])
cv_results = pd.read_excel('/'.join([tables_path, 'cv/20260225/','cv_density_forecasting_20260224.xlsx']))#.drop(columns=["Unnamed: 0"])

cv_results = cv_results[['normalized_density', 'bandwidth', 'kernel', 'robust_std', #'dFPC_dimensions',
                          'KLD', 'JSD', 'L_1', 'L_2', 'L_INFTY']]
cv_results_summary = cv_results.groupby(['normalized_density', 'bandwidth', 'kernel', 'robust_std'], as_index=False).mean()

cv_results_summary["robust_std"] = cv_results_summary["robust_std"].apply(lambda x: True if x == "robust" else False)
cv_results_summary["bandwidth"] = cv_results_summary["bandwidth"].apply(lambda x: "silverman" if x == "rot" else x)

# formatting
## rename columns
corresp_cols =  {
    "kernel": "Kernel",
    "bandwidth": "Bandwidth", 
    "adaptive":  "Adaptive bandwidth", 
    "normalized_density": "Normalized pdf", 
    "robust_std": "Robust $\hat{\sigma}$",
    # "dFPC_dimensions": "Dimensions",
    "L_1": "$L_1$", 
    "L_2": "$L_2$", 
    "L_INFTY": "$L_{\infty}$"
    }
cv_results_summary.rename(columns=corresp_cols, inplace=True)
## sort by metric
cv_results_summary.sort_values(by="KLD", inplace=True)
## ajdust index for viz
cv_results_summary.index = np.arange(1,cv_results_summary.shape[0]+1)
## adjust fild values
# cv_results_summary.loc[:, "kernel"] = cv_results_summary.loc[:, "Kernel"].str.replace("_df_", " + df=").str.replace("t_student", "t-student")
cv_results_summary = cv_results_summary.round(4)

cv_results_summary.head()
```

```python
formatted_cv_summary = (   
    cv_results_summary.iloc[:-2,:]
    .style.background_gradient(
        cmap="RdYlGn_r",   # green = low, red = high
        axis=0            # column-wise
    )
    .format(precision=4)
)
formatted_cv_summary
```

```python
# formatted_cv_summary.hide(axis="index") \
formatted_cv_summary \
    .to_latex(
    buf     = '/'.join([out_path, "cv_results_code.tex"]),
    caption = "Results for expanding window cross-validation t+1 forecasts.",
    label   = "tab:cv_results",
    convert_css=True,
    # column_format="c", #* len(cv_results_summary.columns)
    hrules  = True
    )
```

```python
# Get 3 best models with normalized densities

non_normd_df = cv_results_summary[cv_results_summary["Normalized pdf"]==False]
cv_results_summary_top = pd.concat([
                         cv_results_summary[cv_results_summary["Normalized pdf"]==True].head(3),
                         non_normd_df.head(5),
                         non_normd_df[non_normd_df["Bandwidth"]=="cross-validation"].head(1),
                         non_normd_df[non_normd_df["Robust $\hat{\sigma}$"]==True].head(1),
                         non_normd_df[non_normd_df["Bandwidth"]=="rot"].head(2),
                         non_normd_df[non_normd_df["Kernel"]=="gaussian"].head(2),
                         non_normd_df[non_normd_df["Kernel"]=="epanechnikov"].head(2)
                        ], axis=0) \
                        .drop_duplicates().sort_index()
cv_results_summary_top
```
